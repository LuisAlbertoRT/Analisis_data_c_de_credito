<span style="color:red">**1. En un dataframe (df), ¿cuál es la sintaxis para obtener el total de na’s para cada variable?**</span>


```python
import pandas as pd
datos = {
    'Nombre': ['Pao', 'Alejandro', 'Daniel',"Luis",None,'Pao', 'Alejandro', 'Daniel',"Luis",None,'Pao', 'Alejandro', 'Daniel',"Luis",None],
    'Edad': [21, None, 22,23,24,25, None, 18,17,15,32, None, 13,45,30],
    'Ciudad': ['Madrid', None, 'México',"Turquia","El Salvador",'Madrid', None, 'México',"Turquia","El Salvador",'Madrid', None, 'México',"Turquia","El Salvador"],
    'Nivel_social': ["Alto", None, "Bajo","Alto","Alto","Alto", None, "Bajo","Alto","Bajo","Bajo", None, "Alto","Bajo","Alto"]
}
df = pd.DataFrame(datos)
df.isna().sum()
```

<span style="color:red">**2. En ese mismo df, ¿cuál es la sintaxis para imputar los valores perdidos na con el promedio? Supón que los valores perdidos se encuentran en la variable ‘edad’ y ‘nivel_socieco’.**</span>



```python
#Media redondeada
df['Edad'] = df['Edad'].fillna(round(df['Edad'].mean()))
#Moda
df['Nivel_social'] = df['Nivel_social'].fillna(df['Nivel_social'].mode()[0])
df['Ciudad'] = df['Ciudad'].fillna(df['Ciudad'].mode()[0])
df.head(10)
```

<span style="color:red">**4. Estás a punto de preparar un modelo y para ello quieres separar en entrenamiento y
validación. ¿Con qué sintaxis separarías tu base en un 70% para entrenamiento y el restante

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder



X = df[['Edad', 'Ciudad']]
y = df['Nivel_social']


X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.30, random_state=42
)
```

<span style="color:red">**5. Si quisieras implementar un Random Forest 10 árboles y una máxima profundidad de 2. ¿Cuál sería la sintaxis?**</span>


```python

#Para Y
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_val_encoded = le.transform(y_val)

#Para X
#Continuas
imputer = SimpleImputer(strategy='median')
X_train[['Edad']] = imputer.fit_transform(X_train[['Edad']])
X_val[['Edad']] = imputer.transform(X_val[['Edad']]) 

#Discretas
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
cat_train = pd.DataFrame(
    ohe.fit_transform(X_train[['Ciudad']]), 
    columns=ohe.get_feature_names_out(['Ciudad']),
    index=X_train.index
)
cat_val = pd.DataFrame(
    ohe.transform(X_val[['Ciudad']]), 
    columns=ohe.get_feature_names_out(['Ciudad']),
    index=X_val.index
)


X_train_final = pd.concat([X_train[['Edad']], cat_train], axis=1)
X_val_final = pd.concat([X_val[['Edad']], cat_val], axis=1)


rf = RandomForestClassifier(n_estimators=10, max_depth=2, random_state=42)
rf.fit(X_train_final, y_train_encoded)


predicciones = rf.predict(X_val_final)
print(predicciones)
```

Bajo la regla de negocio se consideraria malo ya que en su revision historica ha tenido tres meses de impago durante un periodo de revision de 12 meses.
El codigo identifica los ultimos 12 meses y hace una comprobacion para encontrar 3 N consecutivas y repite la operacion cada periodo nuevo para evaluar su estado.

Como sugerencia recomiendo no utilizar un clasificador binario si no categorico ya que puede haber hasta 4 veces qun patron de 3N consecutivos por lo que una repeticion de este patron hace al credito aun mas malo. Valdria la pena hacer escalas de que tan malo es.

```python
datos = [
    {
        "monto a pagar": 4447,
        "Ciudad": "MX",
        "Estatus": "Activo",
        "Entidad": 142,
        "frecuencia de pago": "mensual",
        "Número de cuenta": 987690,
        "Fecha apertura": "22/01/20",
        "fecha peor atraso": "30/06/21",
        "fecha actualización": "31/03/22",
        "Comportamiento de pago": ["N", "N", "N", "N", "N", "N", "N", "N", "P", "P", "N", "N", "P"],
        "abierto": 1,
        "activo": 0
    }
]

df = pd.DataFrame(datos)
```

```python
def evaluar_mal_pagador(historial, meses_ventana=12, impagos_consecutivos_malo=3):
    ventana_12m = historial[-meses_ventana:]
    consecutivos = 0
    for pago in ventana_12m:
        if pago == 'N':
            consecutivos += 1
            if consecutivos >= impagos_consecutivos_malo:
                return 1 
        else:
            consecutivos = 0  
    return 0  

df['target_malo'] = df['Comportamiento de pago'].apply(evaluar_mal_pagador)

print(df[['fecha actualización', 'Comportamiento de pago', 'target_malo']])
```

<span style="color:red">**8. Supón que tienes la sig información en las tablas uno y dos: Escribe código sql para hacer un inner join de las tablas y seleccionar la suma de monto *
precio por año.**</span>

```python
!pip install duckdb
```

```python

import duckdb


datos1 = pd.DataFrame({
    'Año': [2012, 2013, 2014],
    'Trim': [3, 4, 1],
    'Precio': [1200, 4000, 100]
})

datos2 = pd.DataFrame({
    'Año': [2012, 2013, 2014],
    'Trim': [3, 4, 1],
    'Monto': [3.2, 5.6, 6.7]
})

query = """
SELECT 
    d1.Año,
    SUM(d2.Monto * d1.Precio) AS Suma
FROM datos1 d1
INNER JOIN datos2 d2 
    ON d1.Año = d2.Año 
   AND d1.Trim = d2.Trim
GROUP BY d1.Año
ORDER BY d1.Año;
"""

resultado = duckdb.query(query).df()
print(resultado)
```
